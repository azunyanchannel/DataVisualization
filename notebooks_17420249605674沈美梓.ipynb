```python
import numpy as np

# ----------------------------------------------------
# 1. 创建一个3x4维的二维数组，其中所有元素都为0，并打印该数组
# ----------------------------------------------------
arr1 = np.zeros((3, 4))
print("1. 3x4维的全0数组：")
print(arr1)
print("-" * 40)

# ----------------------------------------------------
# 2. 创建一个包含从0到9所有整数的一维数组，然后将其重塑(reshape)为一个2x5的二维数组
# ----------------------------------------------------
arr2 = np.arange(10).reshape(2, 5)
print("2. 重塑后的2x5二维数组：")
print(arr2)
print("-" * 40)

# ----------------------------------------------------
# 3. 将以下2x3的数组变换为一个3x2的数组
# arr = np.array([[1, 2, 3], [4, 5, 6]])
# ----------------------------------------------------
arr = np.array([[1, 2, 3], [4, 5, 6]])
arr3 = arr.reshape(3, 2)
print("3. 变换后的3x2数组：")
print(arr3)
print("-" * 40)

# ----------------------------------------------------
# 4. 计算上一题中的二维数组 arr 每一列的平均值
# ----------------------------------------------------
col_mean = np.mean(arr, axis=0)
print("4. 每一列的平均值 (axis=0)：")
print(col_mean)
print("-" * 40)

# ----------------------------------------------------
# 5. 计算上一题中的二维数组 arr 每一行的标准差
# ----------------------------------------------------
row_std = np.std(arr, axis=1)
print("5. 每一行的标准差 (axis=1)：")
print(row_std)
print("-" * 40)
```
