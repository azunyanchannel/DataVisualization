### DataFrame数据选取

####  `loc[]`：基于行索引和列索引(标签)的方法选择数据

```python
df.loc[:,'order_id'] #（1）选择指定标签对应的列
type(df.loc[:,'order_id']) #取出的数据是一维数据
```

```python
df.loc[:,['order_id']] #（2）选择指定标签对应的列
type(df.loc[:,['order_id']]) #取出的数据是二维表格
```

```python
df.loc[:,['order_id', 'order_date']] #选取多列数据
```

#### `iloc[]`: 基于行和列的下标位置选择数据

```python
df.iloc[0] # 选择第一行，等价于df.iloc[0,:]，得到的是Series
```

```python
df.iloc[[0]] # 选择第一行，得到的是DataFrame
```

```python
df.iloc[1:,] # 选择从第二行开始的全部数据
```

```python
df.iloc[:,[0]] # 选择第一列数据
```

```python
df.iloc[:,[0,1]] # 选择前两列数据
```

```python
df.iloc[0,0] # 基于行和列的位置选取特定数据
```

```python
#直接选取某列或若干列的数据
# df[['order_id']]
# df.order_id
df[['order_id','order_date']]
```

```python
#选取符合条件的数据
df[df.amount>10] # 等价于df.loc[df.amount>10]
```

```python
# dataframe的连接
# 以字典的形式定义df1
df1 = pd.DataFrame({'alpha':['A','B','B','C','D','E'],'feature1':[1,1,2,3,3,1],
            'feature2':['low','medium','medium','high','low','high']})

df1.head()
```

```python
# 定义df2
df2 = pd.DataFrame({'alpha':['A','A','B','F'],'pazham':['apple','orange','pine','pear'],
            'kilo':['high','low','high','medium'],'price':np.array([5,6,5,7])})
df2.head()
```

```python
df3 = pd.concat([df1,df2], axis=1) #按列拼接
df3
```

```python
# 删除缺失值
# df3.isna()
df3.dropna()
```

```python
pd.concat([df1,df2], axis=0) #按行拼接
```

```python
df1.merge(df2,how='inner',on='alpha') # 基于共同列alpha的内连接
```

```python
df1.merge(df2,how='outer',on='alpha') # 基于共同列的并集进行外连接
```

```python
df1.merge(df2,how='left',on='alpha') # 基于共同列alpha的左连接
```

```python
df1.merge(df2,how='right',on='alpha')# 基于共同列alpha的右连接
```
