## 2. Matplotlib多幅子图与坐标系的添加方法 

### `方法一：` plt.subplot方式依次添加 

````python
plt.style.use('default') # 使用默认样式

plt.figure(figsize=(8,6))
plt.subplot(221,facecolor='#FCE38A')
plt.subplot(222,facecolor='#EAFFD0')
plt.subplot(212,facecolor='#95E1D3')
plt.show()
````

````python
# plt.rcParams['axes.unicode_minus']=False # 解决负号显示不出来的问题

plt.figure(figsize=(8,6))
X = np.arange(0.01, 10, 0.01)
plt.subplot(221,facecolor='#FCE38A')
plt.plot(X, np.sin(X), 'b') # 绘制正弦函数

plt.subplot(222,facecolor='#EAFFD0')
plt.plot(X, np.cos(X), 'g') # 绘制余弦函数

plt.subplot(212,facecolor='#95E1D3')
plt.bar(np.arange(6), np.array([2, 4, 1, 6, 3, 8]), color='#eaeaea') # 绘制柱状图

# 绘制一个居中的总标题
plt.suptitle('Pyplot') # 注意和plt.title()进行区分

# 保存图形到本地
plt.savefig('pyplot.jpg',format='jpg',dpi=200)

# 设置字体
plt.rcParams['font.family']='sans-serif'

plt.show()
````

### `方法二：` add_subplot方式依次添加
#### 参数设置同plt.subplot一致

````python
fig = plt.figure()

# 依次在各个子图上进行添加
ax1 = fig.add_subplot(221,facecolor='#2B9ACE')
ax1.plot(X, np.sin(X), 'w')

ax2 = fig.add_subplot(222,facecolor='#89BD54')
ax2.plot(X, np.cos(X), 'y')

ax3 = fig.add_subplot(212,facecolor='#E6E243')
ax3.bar(np.arange(6), np.array([2, 4, 1, 6, 3, 8]),color='c')

fig.suptitle('add_subplot')
plt.show()
````

### `方法三：` subplots
#### 相比subplot和add_subplot，函数调用比较简洁，但是不能自定义子图布局

````python
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(ncols=2, nrows=2, figsize=(8, 6))
ax1.set_facecolor("#8F4B99")
ax2.set_facecolor("#89BD54")
ax3.set_facecolor("#E6E243")
ax4.set_facecolor("#2B9ACE")
fig.suptitle('subplots')
plt.show()
````

````python
fig, ax = plt.subplots(ncols=2, nrows=2, figsize=(8, 6))

ax[0,0].set_facecolor("#8F4B99")
ax[0,1].set_facecolor("#89BD54")
ax[1,0].set_facecolor("#E6E243")
ax[1,1].set_facecolor("#2B9ACE")
fig.suptitle('subplots')
plt.show()
````

````python
fig, ax = plt.subplots(ncols=2, nrows=2, figsize=(8, 6), sharex=True)
X = np.arange(0.01, 10, 0.01)
ax[0,0].plot(X, 2*X - 1)
ax[0,0].set_facecolor("#8F4B99")

ax[0,1].plot(X, np.log(X))
ax[0,1].set_facecolor("#89BD54")

ax[1,0].plot(X, np.exp(X))
ax[1,0].set_facecolor("#E6E243")

ax[1,1].plot(X, np.sin(X))
ax[1,1].set_facecolor("#2B9ACE")

fig.suptitle('subplots')
plt.show()
````

### 3.1 线图 plot() 

````python
mpl.rcParams['font.sans-serif']=['SimHei'] # 用来正常显示中文，字体设置为黑体
mpl.rcParams['axes.unicode_minus']=False # 用来正常显示负号

plt.style.use('default') 

# 读取数据集MappingAnalysis_Data.csv
df=pd.read_csv("MappingAnalysis_Data.csv",index_col=0)
print(df.head())
print(df.variable.unique())

df1=df[df.variable=='0%(Control)']
df2=df[df.variable=='1%']
df3=df[df.variable=='5%']
df4=df[df.variable=='15%']

# =============================================================================
colors=['#e41a1c','#377eb8','#4daf4a','#984ea3'] # 定义颜色
markers=['o','s','H','D']  # 定义标记  
labels=["0%(Control)","1%","5%","15%"] # 定义标签
````

````python
#----------------------------(a) 绘制线图 -----------------------------------------------
fig =plt.figure(figsize=(4,3), dpi=100)

for i in range(0,4):
    temp_df=df[df.variable==labels[i]]
    plt.plot(temp_df.Time, temp_df.value)
    
plt.show()
````

````python
#-------------------------------(b) 绘制带标记和图例的线图 --------------------------------------------
fig =plt.figure(figsize=(4,3), dpi=100)

for i in range(0,4):
    plt.plot(df[df.variable==labels[i]].Time, df[df.variable==labels[i]].value, 
         marker=markers[i], markerfacecolor=colors[i], markersize=6, markeredgewidth=0.5,
         color="k", linewidth=0.5, linestyle="-",label=labels[i])

plt.legend(loc='upper left') # 设置图例的显示位置
plt.show()
````

````python
#---------------------------------(c) 更多个性化设置 ------------------------------------------
fig =plt.figure(figsize=(4,3), dpi=100)

plt.plot(df1.Time, df1.value, 
         marker=markers[0], markerfacecolor=colors[0], markeredgecolor=colors[0], markersize=8, markeredgewidth=0.5,
         color="k", linewidth=0.5, linestyle="-",label=labels[0])
plt.plot(df2.Time, df2.value, 
         marker=markers[1], markerfacecolor=colors[1], markeredgecolor=colors[1], markersize=7, markeredgewidth=0.5,
         color="k", linewidth=0.5, linestyle="-",label=labels[1])
plt.plot(df3.Time, df3.value, 
         marker=markers[2], markerfacecolor=colors[2], markeredgecolor=colors[2], markersize=8, markeredgewidth=0.5,
         color="k", linewidth=0.5, linestyle="-",label=labels[2])
plt.plot(df4.Time, df4.value, 
         marker=markers[3], markerfacecolor=colors[3], markeredgecolor=colors[3], markersize=7, markeredgewidth=0.5,
         color="k", linewidth=0.5, linestyle="-",label=labels[3])

# 设置图例         
plt.legend(loc='upper left',edgecolor='none',facecolor='none')

# 设置横轴和纵轴的名称
plt.xlabel("Time(d)",fontsize=14)
plt.ylabel("Value",fontsize=14)

# 设置坐标轴范围
plt.xlim(-1,20)
plt.ylim(-2,90)

# 设置横轴刻度, np.linespace()生成指定范围内指定个数的均等分布
plt.xticks(np.linspace(0,20,11,endpoint=True),fontsize=10)
# 设置纵轴刻度
plt.yticks(np.linspace(0,90,10,endpoint=True),fontsize=10)

# 删除右边和顶部的绘图区域边框线。注意：子图的边框线需要在子图层面进行操作
ax = plt.gca() # 返回当前的子图对象
ax.spines['right'].set_color('none') # 取消坐标系右侧的颜色，或者ax.spines['right'].set_visible(False)
ax.spines['top'].set_color('none') # 取消坐标系上方的颜色
plt.show()
````

### 3.3 箱型图 boxplot()、小提琴图 violinplot()

````python
plt.style.use('tableau-colorblind10') # 更改样式

fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(13, 4),sharey=True)

# 设置种子，确保结果的可复制性
np.random.seed(19680801)

# 从四个均值为0，标准差分别为6、7、8、9的正态分布中随机抽取100个样本
all_data = [np.random.normal(0, std, 100) for std in range(6, 10)]

# (1) plot box plot
axs[0].boxplot(all_data,notch=True,sym='o') # sym表示异常值的符号 sym='b+'
axs[0].set_title('Box Plot')

# (2) plot violin plot
axs[1].violinplot(all_data,showmeans=False,showmedians=True,showextrema=True)
axs[1].set_title('Default Violin Plot')

# (3) plot customized violin plot
parts = axs[2].violinplot(all_data,showmeans=False,showmedians=True,showextrema=True)
axs[2].set_title('Customized Violin Plot')
for pc in parts['bodies']:
    pc.set_facecolor('#FFC7C7') # 设置背景色
    pc.set_edgecolor('#8785A2') # 设置边界颜色
    pc.set_alpha(0.7) # 设置透明度

# 生成分位数：Q1, Median, Q3
quartile1, medians, quartile3 = np.percentile(all_data, [25, 50, 75], axis=1)

inds = np.arange(1, len(medians) + 1) # X轴坐标是1,2,3,4
axs[2].scatter(inds, medians, marker='o', color='w', s=20, zorder=3) # 以散点图的形式增加中位数
axs[2].vlines(inds, quartile1, quartile3, color='k', linestyle='-', lw=8) # 以竖直线的形式增加Q1, Q3

# 设置刻度、网格线、坐标轴标签
for ax in axs:
    ax.yaxis.grid(True)
    ax.set_xticks([y + 1 for y in range(len(all_data))]) 
    ax.set_xlabel('Four separate samples')
    ax.set_ylabel('Observed values')

plt.rcParams['font.family'] = 'sans-serif'
plt.show()
````

### 3.4 散点图 scatter()

### 3.5 密度图 hist() 

$\Large f(x|\mu,\sigma^2)=\frac {1}{\sqrt {2\pi\sigma^2}} \exp({-\frac {(x-\mu)^2}{2\sigma^2}})$

````python
# 设置种子
np.random.seed(19680801)

# 生成数据
mu = 200 
sigma = 25
n_bins = 25 # 分箱数
data = np.random.normal(mu, sigma, size=100)

fig,ax = plt.subplots(figsize=(6, 5),dpi=100)

# ECDF
n, bins, patches = ax.hist(data, n_bins, density=True, histtype="step", 
                               cumulative=True, label="Empirical Cumulative Distribution Histogram") # histtype='stepfilled'

x = np.linspace(data.min(), data.max())
y = ((1 / (np.sqrt(2 * np.pi) * sigma)) *
     np.exp(-0.5 * (1 / sigma * (x - mu))**2)) # PDF
y = y.cumsum() # 求累积密度
y /= y[-1] # 进行标准化，使累积密度的最大值为1
ax.plot(x, y, "k--", linewidth=1.5, label="Theory")

fig.suptitle("Cumulative Distributions")

ax.grid(True)
ax.legend(loc='upper left')
ax.set_xlabel("Annual rainfall (mm)")
ax.set_ylabel("Probability of occurrence")

plt.show()
````
