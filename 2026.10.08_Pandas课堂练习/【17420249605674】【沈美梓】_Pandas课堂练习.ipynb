```python
import pandas as pd
import numpy as np

# 设置 pandas 显示选项，便于查看完整表格
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 1000)
```

### 1) 读取数据到 DataFrame 中

```python
# 1) 读取 customers.txt 数据集
df = pd.read_csv('customers.txt')
print("数据集总维度 (行数, 列数)：", df.shape)
df.head()
```

### 2) 查看数据集中各个属性（列）的类型

```python
# 2) 查看数据集中各个属性（列）的数据类型
print("各属性数据类型：")
print(df.dtypes)
```

### 3) 打印前五行和后五行的数据

```python
# 3) 打印前五行和后五行的数据
print("====== 前五行数据 (head) ======")
display(df.head(5))

print("\n====== 后五行数据 (tail) ======")
display(df.tail(5))
```

### 4) 查看属性 education、occupation、income 的可能取值有哪些

```python
# 4) 查看指定属性的可能取值（唯一值）
target_columns = ['education', 'occupation', 'income']

for col in target_columns:
    unique_vals = df[col].unique()
    print(f"【{col}】的可能取值 ({len(unique_vals)} 种)：")
    print(list(unique_vals))
    print("-" * 60)
```

### 5) 选择出客户 ID 为 "Cust-21910" 的行

```python
# 5) 通过布尔索引筛选出客户 ID 为 'Cust-21910' 的行
cust_21910 = df[df['cust_id'] == 'Cust-21910']
cust_21910
```

### 6) 将 marital 字段重新赋值：是改为 1，否改为 0

```python
# 6) 将 marital 字段中的 '是' 映射为 1，'否' 映射为 0
df['marital'] = df['marital'].map({'是': 1, '否': 0})

# 验证赋值结果与分布
print("marital 字段唯一值：", df['marital'].unique())
print("重新赋值后的 marital 频数统计：")
print(df['marital'].value_counts())
```

### 7) 将数据拆分为男性客户和女性客户两个样本

```python
# 7) 按性别 (gender) 拆分样本
male_df = df[df['gender'] == '男'].copy()
female_df = df[df['gender'] == '女'].copy()

print(f"男性客户样本数量：{len(male_df)} 人")
print(f"女性客户样本数量：{len(female_df)} 人")
print(f"总计验证：{len(male_df) + len(female_df)} / {len(df)}")
```

### 8) 依次输出男性客户中，高价值客户、中等价值客户、低价值客户的教育水平的众数、已婚人数比例，以及年龄的均值

```python
# 8) 分析男性客户在各价值等级下的特征指标
# 注：数据集中 '一般价值客户' 即对应题目中的 '中等价值客户'
customer_categories = ['高价值客户', '一般价值客户', '低价值客户']

summary_records = []

for cat in customer_categories:
    sub_df = male_df[male_df['custcat'] == cat]
    
    # 1. 教育水平众数
    edu_mode = sub_df['education'].mode()[0]
    # 2. 已婚人数比例 (marital为1的比例，即mean)
    married_ratio = sub_df['marital'].mean()
    # 3. 年龄均值
    age_mean = sub_df['age'].mean()
    
    # 格式化名称便于对应题目
    display_name = '中等价值客户 (一般价值客户)' if cat == '一般价值客户' else cat
    print(f"【{display_name}】(样本数: {len(sub_df)})：")
    print(f"  - 教育水平众数: {edu_mode}")
    print(f"  - 已婚人数比例: {married_ratio:.4f} ({married_ratio*100:.2f}%)")
    print(f"  - 年龄均值:     {age_mean:.2f} 岁")
    print("-" * 50)
    
    summary_records.append({
        '客户分类': display_name,
        '样本人数': len(sub_df),
        '教育水平众数': edu_mode,
        '已婚比例': f"{married_ratio*100:.2f}%",
        '年龄均值': round(age_mean, 2)
    })

# 输出格式化汇总表
summary_df = pd.DataFrame(summary_records)
display(summary_df)
```

### 9) 将女性客户样本导出到 Excel 中

```python
# 9) 将女性客户样本数据导出为 Excel 文件
output_excel = '女性客户样本.xlsx'
female_df.to_excel(output_excel, index=False)
print(f"已成功将女性客户样本导出到 Excel 文件：{output_excel}")
print(f"导出记录总数：{len(female_df)} 行，字段数：{female_df.shape[1]} 列")
```
